```python
#video conversion of Tiff multi-image files to mp4 or gif for visualization
import tifffile
import numpy as np
import imageio
import os
```

```python
def convert_video(input_file, output_filename):
    tf = tifffile.TiffFile(input_file)

    images = []

    for page in tf.pages:
        image = page.asarray()
        sampled_im = image[::2, ::2]
        sampled_im = sampled_im[100:350,100:350]
        # quantized_image = sampled_im.astype(np.uint8)
        quantized_image = (sampled_im/65535*255).astype(np.uint8)

        images.append(quantized_image)

    print(len(images))
    imageio.mimsave(output_filename, images, loop=0)
```

```python
base_dir = '/'
run = ''
subfolder = base_dir + run + '/'
```

```python
for root, dirs, files in os.walk(subfolder):
    for file in files:
        if '.tif' in file:
            file_path = os.path.join(root, file)
            output_path = root + '/' + file[:-4] + '.gif'
            print('converting ',root,file)
            convert_video(file_path, output_path)
```
